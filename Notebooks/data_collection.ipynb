# 🧠 Stock Scoring Algorithm – Complete Technical Documentation

## 1. Overview

The scoring engine evaluates stocks across **six fundamental & technical categories**, applies **risk adjustments**, and produces a **final score (0–100)** with a **textual recommendation**. The system is designed to be:

- **Modular** – each scorer is independent
- **Adaptive** – sector‑aware normalisation ranges
- **Robust** – handles missing data gracefully (returns neutral 50)
- **Configurable** – weights and sector overrides are centralised

**Pipeline flow:**

```
Raw data (Yahoo Finance)
    ↓
MetricStandardizer (unit conversion)
    ↓
6 Category Scorers (valuation, profitability, growth, financial health, dividend, technical)
    ↓
Weighted sum (based on stock type)
    ↓
RiskAdjustments (penalties / bonuses)
    ↓
Final score & recommendation
```

---

## 2. Data Schema (Input)

The scoring engine expects a dictionary with the following top‑level keys:

| Key | Type | Description |
|-----|------|-------------|
| `info` | dict | Fundamental & valuation metrics (see list below) |
| `history` | dict | Price/volume history + computed indicators (MA, RSI, etc.) |
| `dividends` | dict | Dividend history series + pre‑computed metrics (years, cuts, CAGR) |
| `financials` | dict | Quarterly financials DataFrame + derived metrics (`positive_quarters`, `profitable_quarters`) |

All `info` fields are standardised by `MetricStandardizer` to consistent units:

- **Percentages** → stored as real numbers (e.g., `profit_margin = 25` for 25%)
- **Ratios** → stored as raw ratios (e.g., `debt_to_equity = 0.8`)
- **Dividend yield** → stored as decimal (e.g., `0.0038` for 0.38%)

---

## 3. Category Scorers (6 components)

Each scorer returns a `{"score": float, "details": {...}}` dictionary.

### 3.1 ValuationScorer (inverse normalisation – lower multiple = higher score)

| Metric | Weight | Normalisation range (default) | Sector‑aware |
|--------|--------|------------------------------|--------------|
| P/E ratio | 30% | 5 – 60 | ✅ (e.g., Technology 5–70) |
| PEG ratio | 25% | 0.5 – 3 | ✅ |
| Price/Book | 20% | 0.5 – 10 | ✅ (Technology 0.5–20) |
| Price/Sales | 15% | 0.5 – 15 | ✅ |
| EV/EBITDA | 10% | 3 – 30 | ✅ |

**Formula:** `inverse_score = 100 - ((value - min) / (max - min)) * 100`  
Values below min → 100, above max → 0.  
If value ≤ 0 (e.g., negative earnings) → score = 0.

---

### 3.2 ProfitabilityScorer (linear normalisation)

| Metric | Weight | Default range (min–max) |
|--------|--------|--------------------------|
| ROE (Return on Equity) | 30% | 0% – 30% |
| ROA (Return on Assets) | 20% | 0% – 20% |
| Profit margin | 25% | 0% – 30% |
| Operating margin | 15% | 0% – 35% |
| Gross margin | 10% | 0% – 60% |

**Formula:** `linear_score = ((value - min) / (max - min)) * 100`  
Values below min → 0, above max → 100.

Sector‑aware overrides: Technology uses ROE 0–40%, ROA 0–25%, etc.

---

### 3.3 GrowthScorer

| Metric | Weight | Normalisation range (min–max) |
|--------|--------|------------------------------|
| Revenue growth (YoY) | 30% | -20% – +40% |
| Earnings growth (YoY) | 30% | -20% – +40% |
| EPS growth (forward vs trailing) | 20% | -20% – +40% |
| Consistency (positive growth quarters, 0–8) | 20% | 0 – 8 |

All use **linear normalisation**. Missing values default to 50.

---

### 3.4 FinancialHealthScorer

| Metric | Weight | Normalisation | Default range |
|--------|--------|---------------|----------------|
| Debt/Equity | 30% | inverse | 0 – 3 (ratio) |
| Current ratio | 30% | linear | 0.5 – 4 |
| Quick ratio | 20% | linear | 0.2 – 3 |
| Free Cash Flow yield | 20% | linear | -5% – +15% |

**Sector overrides:** Technology uses current ratio 0.5–2.5, quick 0.2–2.0.

---

### 3.5 DividendScorer

| Metric | Weight | Normalisation | Range |
|--------|--------|---------------|-------|
| Dividend yield | 30% | linear | 0% – 10% |
| Payout ratio | 30% | inverse | 0% – 150% |
| Dividend CAGR (5 years) | 40% | linear | -10% – +20% |

**Penalty:** If `has_cuts = True`, dividend growth score is halved.

---

### 3.6 TechnicalScorer

| Metric | Weight | Logic |
|--------|--------|-------|
| MA alignment | 40% | price > MA50 > MA200 → 100; price > MA200 → 70; else 30; default 50 |
| RSI (14) | 30% | linear distance from 50: `100 - abs(RSI-50)*2` (clamped 0–100) |
| MACD | 20% | 100 if MACD > signal line, else 40; neutral 50 |
| Volume ratio (current / 20d avg) | 10% | linear normalisation 0.5–2 (0.5→0, 2→100) |

---

## 4. Stock Type Weights

Weights are defined in `weights.py`:

| Category | Growth Stock | Value Stock | Blend Stock |
|----------|--------------|-------------|-------------|
| Valuation | 15% | 30% | 25% |
| Profitability | 20% | 25% | 25% |
| Growth | 35% | 10% | 20% |
| Financial Health | 15% | 15% | 15% |
| Dividend | 5% | 15% | 10% |
| Technical | 10% | 5% | 5% |

---

## 5. Risk Adjustments (`RiskAdjustments.apply`)

These are **modifiers** (positive or negative) applied after the weighted sum.

| Condition | Adjustment |
|-----------|------------|
| Market cap < $2B (small cap) | -5 |
| Market cap > $200B (mega cap) | +2 |
| Beta > 1.5 (high volatility) | -3 |
| Beta < 0.5 (very low volatility) | +2 |
| Sector in high‑risk list (Biotech, Crypto, Cannabis, Energy, Basic Materials) | -3 |
| Payout ratio > 1 and Free Cash Flow negative | -4 |
| P/E < 10 and earnings growth negative (value trap) | -5 |
| MA50 < MA200 (weak momentum) | -2 |
| Free Cash Flow negative | -3 |

Final score is **clipped to 0–100**.

---

## 6. Sector‑Aware Configuration

Located in `config/sector_config.py`.  
Defines `DEFAULT_RANGES` and `SECTOR_RANGES` for metrics like:

- `price_to_book`, `current_ratio`, `roe`, etc.
- If a metric is set to `None` for a sector, it is skipped (score = 50).

**Supported sectors:** Technology, Financial Services, Real Estate, Energy, Healthcare, Consumer Cyclical, Consumer Defensive, Industrials, Utilities, Basic Materials, Communication Services, Crypto, Biotechnology, Cannabis.

---

## 7. Utility Modules

### MetricStandardizer
- Converts Yahoo Finance raw data to internal units:
  - Decimal → percent for margins and returns (×100)
  - Debt/Equity from Yahoo’s 79 → 0.79 (÷100)
  - Dividend yield from % → decimal (÷100)
  - Removes absurd payout ratios (>1000% → None)

### Normalisation functions (`normalization.py`)
- `clamp(value, min, max)`
- `normalize_linear(value, min, max)` → returns 0–100, missing → 50
- `normalize_inverse(value, min, max)` → returns 0–100 (higher score = lower raw value), missing → 50

### safe_math
- `safe_float(value)` – converts to float, returns `None` for NaN, inf, or conversion errors.

## 8. Output & Recommendation

**Final score mapping to recommendation:**

| Score range | Recommendation |
|-------------|----------------|
| 85 – 100 | Strong Buy |
| 70 – 84  | Buy |
| 55 – 69  | Hold |
| 40 – 54  | Weak Hold |
| 0 – 39   | Sell |

**Output dictionary from `StockScorer.score()`:**

```python
{
    "stock_type": "Blend Stock",
    "category_scores": { ... },   # each with score, details, weight, weighted_score
    "raw_score": 57.84,
    "final_score": 59.84,
    "recommendation": "Hold"
}
```

---

## 9. Example: Apple (AAPL) – Real Output

- Final score: 59.84 → **Hold**
- Strong profitability (94.4) and technical (81.6)
- Weak valuation (25.1) due to high P/B and P/S
- Low dividend yield (score 0.04) drags dividend category
- Sector‑aware ranges improved liquidity scores (current ratio 16 → 28.5)

## 10. Adding a New Scorer

1. Subclass `BaseScorer`
2. Implement `calculate(self)` returning `{"score": float, "details": {...}}`
3. Add to `StockScorer.score()` and include weight in `WEIGHTS`
4. (Optional) Add sector overrides to `sector_config.py`

---

## 11. Limitations & Future Enhancements

- **Trailing‑only** – no forward adjustments (could be added via `RiskAdjustments`)
- **Fixed ranges** – future version could use percentiles across a universe
- **No earnings surprise or analyst revision metrics**
- **Dividend growth may be unreliable for stocks with short history**
